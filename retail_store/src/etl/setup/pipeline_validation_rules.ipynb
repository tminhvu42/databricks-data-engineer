```python
dbutils.widgets.text("catalog_name", "")
dbutils.widgets.text("schema_name", "")

catalog_name = dbutils.widgets.get("catalog_name")
schema_name = dbutils.widgets.get("schema_name")
```

```python
spark.sql(f"""CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema_name}""")
```

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {catalog_name}.{schema_name}.pipeline_validation_rules (
    pipeline_name STRING,
    dataset_name STRING,
    rule_name STRING,
    sql_constraint STRING,
    tags ARRAY<STRING>
) USING DELTA;
""")

## Populating validation metadata
spark.sql(f"""
INSERT INTO {catalog_name}.{schema_name}.pipeline_validation_rules VALUES 
('location-etl', 'silver_location', 'valid_id', 'location_id IS NOT NULL', array('validity', 'location')),
('location-etl', 'silver_location', 'valid_email', 'email LIKE "%@%.%"', array('validity', 'location')),
('employees-etl', 'silver_employees', 'valid_id', 'employee_id IS NOT NULL', array('validity', 'employees')),
('employees-etl', 'silver_employees', 'valid_email', 'email LIKE "%@%.%"', array('validity', 'employees'));
""")
```
